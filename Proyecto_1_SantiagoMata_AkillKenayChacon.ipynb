## Análisis de los datos integrados

A partir de la integración de las tres fuentes se comparan los indicadores educativos y socioeconómicos entre las zonas urbana y rural durante 2021 y 2022.

### Rendimiento académico entre las zonas urbana y rural

Para comparar el rendimiento académico se utilizan los porcentajes de estudiantes aprobados y aplazados de cada zona durante 2021 y 2022.

```python
comparacion_rendimiento = datos_finales[
    ["Año", "Zona", "Porcentaje_aprobados", "Porcentaje_aplazados"]
]

comparacion_rendimiento
# selecciono las variables necesarias para comparar el rendimiento entre zonas
```

```python
aprobados_zona = comparacion_rendimiento.set_index(
    ["Año", "Zona"]
)["Porcentaje_aprobados"].unstack()

aprobados_zona
# organizo los porcentajes de aprobados para comparar ambas zonas por año
```

```python
aprobados_zona["Diferencia"] = (
    aprobados_zona["Rural"] - aprobados_zona["Urbana"]
)

aprobados_zona
# calculo la diferencia en puntos porcentuales entre la zona rural y urbana
```

```python
aplazados_zona = comparacion_rendimiento.set_index(
    ["Año", "Zona"]
)["Porcentaje_aplazados"].unstack()

aplazados_zona
# organizo los porcentajes de aplazados para comparar ambas zonas por año
```

```python
aplazados_zona["Diferencia"] = (
    aplazados_zona["Urbana"] - aplazados_zona["Rural"]
)

aplazados_zona
# calculo la diferencia en puntos porcentuales de aplazados entre la zona urbana y rural
```

### Interpretación del rendimiento académico

Durante 2021 y 2022 la zona rural presentó un porcentaje de aprobación mayor que la zona urbana. En 2021 la diferencia fue de aproximadamente 1.17% y en 2022 aumentó a 1.61%.

Por otra parte, la zona urbana presentó un mayor porcentaje de estudiantes aplazados. La diferencia fue de aproximadamente 1.16% en 2021 y 1.57% en 2022.

Por lo tanto, durante los dos años analizados se observa un rendimiento académico ligeramente más positivo en la zona rural según los porcentajes de aprobación y aplazamiento.

### Interpretación de los indicadores de empleo y pobreza

Durante 2021 y 2022 la zona urbana presentó mayores tasas de participación y ocupación que la zona rural. La diferencia en participación fue de aproximadamente 4.84% en 2021 y 4.20% en 2022, mientras que la diferencia en ocupación pasó de 2.91% a 3.32%.

En cuanto al desempleo, la zona urbana presentó una tasa mayor en ambos años. Sin embargo, la diferencia entre las zonas disminuyó de aproximadamente 1.99% en 2021 a 0.67% en 2022. Además, la tasa de desempleo disminuyó en ambas zonas durante este periodo.

Por otra parte, la pobreza fue mayor en la zona rural. La diferencia con respecto a la zona urbana fue de aproximadamente 4.58% en 2021 y aumentó a 7.21% en 2022. Mientras la pobreza rural aumentó entre ambos años, la pobreza urbana disminuyó ligeramente.

En general, los datos muestran diferencias socioeconómicas entre ambas zonas: la zona urbana presentó mayores niveles de participación y ocupación, mientras que la zona rural presentó mayores niveles de pobreza.
